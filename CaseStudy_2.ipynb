PART A

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.integrate import quad

data = {
    'Student': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6', 'S7', 'S8', 'S9', 'S10'],
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    'Attendance': [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    'Assignment_Score': [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    'Previous_Score': [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    'Final_Score': [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}

df = pd.DataFrame(data)

print("Number of students:", len(df))

print("\nColumn Names:", df.columns.tolist())

print("\nData Types:\n", df.dtypes)

print("\nFirst 5 Records:\n", df.head())

print("\nLast 5 Records:\n", df.tail())
```

PART B

```python
print("Complete Dataset:\n", df)

averages = df[['Study_Hours', 'Attendance', 'Assignment_Score', 'Previous_Score', 'Final_Score']].mean()
print("\nAverages:\n", averages)

max_student = df.loc[df['Final_Score'].idxmax()]
print(f"\nHighest Final Score: {max_student['Student']} ({max_student['Final_Score']})")

min_student = df.loc[df['Final_Score'].idxmin()]
print(f"Lowest Final Score: {min_student['Student']} ({min_student['Final_Score']})")

print("\nStudents with Attendance > 80%:\n", df[df['Attendance'] > 80][['Student', 'Attendance']])

print("\nStudents with Final Score > 80:\n", df[df['Final_Score'] > 80][['Student', 'Final_Score']])

print("\nStatistical Summary:\n", df.describe())
```

PART C

```python
final_scores=df['Final_Score'].to_numpy()

mean_val=np.mean(final_scores)
median_val=np.median(final_scores)
max_val=np.max(final_scores)
min_val=np.min(final_scores)
std_val=np.std(final_scores, ddof=0)

print(f"Mean: {mean_val:.2f}")
print(f"Median: {median_val:.2f}")
print(f"Maximum: {max_val}")
print(f"Minimum: {min_val}")
print(f"Standard Deviation: {std_val}")
```

```python
X=df[["Study_Hours", "Attendance", "Assignment_Score"]].values
print("Shape of X:", X.shape)
X_T=X.T
XTX=X_T@X
study_hours=df["Study_Hours"].values
final_scores=df["Final_Score"].values
dot_product=np.dot(study_hours, final_scores)
print("\nMatrix X(first 3 rows):\n",X[:3])
print("\nShape of X^T:", X_T.shape)
print("\nX^T * X Matrix(3x3):\n", XTX)
print("\nDot Product (Study Hours . Final Score):", dot_product)
```

PART I

```python
X_raw=df[["Study_Hours","Attendance","Assignment_Score"]].values
Y=df["Final_Score"].values

X_design=np.column_stack((np.ones(len(X_raw)), X_raw))

beta=np.linalg.inv(X_design.T@X_design)@X_design.T@Y
print("Intercept(b0):", round(beta[0],4))
print("Study Hours Coeff(b1):", round(beta[1],4))
print("Attendance Coeff(b2):", round(beta[2],4))
print("Assignment Score Coeff(b3):", round(beta[3],4))
```
